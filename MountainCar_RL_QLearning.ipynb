# Projet AML — MountainCar & Reinforcement Learning

**Q-Learning tabulaire et Deep RL**

- Cours : Advanced Machine Learning (AML) — ESPRIT, 4DAS — Projet 3 : Q-Learning
- Article de référence : *Modelling a Reinforcement Learning Agent for Mountain Car Problem Using Q-Learning With Tabular Discretization* (MysuruCon, 2022) — DOI 10.1109/mysurucon55714.2022.9972352
- Environnement : Gymnasium `MountainCar-v0` (et `MountainCarContinuous-v0` pour TD3)
- Méthodologie : **TDSP** (Team Data Science Process)

## 1. Business Understanding

### 1.1 Contexte

Un **agent** apprend à décider en interagissant avec un **environnement** : à chaque étape il observe un état, choisit une action et reçoit une récompense. Il n'y a pas de dataset étiqueté, l'agent apprend par essai-erreur.

**MountainCar** : une voiture au fond d'une vallée doit atteindre le drapeau au sommet de la colline de droite. Son moteur est trop faible : elle doit d'abord reculer pour prendre de l'élan.

| Élément | Description |
|---|---|
| Observation | position ∈ [−1,2 ; 0,6], vitesse ∈ [−0,07 ; 0,07] |
| Actions | 0 = gauche, 1 = rien, 2 = droite |
| Récompense | −1 à chaque étape |
| Fin (terminated) | position ≥ 0,5 |
| Troncature (truncated) | 200 étapes |

**Difficultés** : récompense rare, action contre-intuitive, espace d'états continu (discrétisation nécessaire pour le Q-Learning), compromis exploration / exploitation.

### 1.2 Problématique

> Comment un agent peut-il apprendre, uniquement par essai-erreur, une stratégie efficace et fiable pour un problème de contrôle à récompense rare — et dans quelle mesure le Deep RL apporte-t-il un gain réel par rapport au Q-Learning tabulaire sur un problème de faible dimension ?

Questions de recherche : (1) **reproductibilité** de l'article, (2) **amélioration** du Q-Learning, (3) **nouvelles approches** (DQN, PPO, TD3), (4) **compromis** performance / stabilité / temps / ressources.

### 1.3 Objectifs métier (BO) et data science (DSO)

| BO | Objectif métier | DSO | Objectif data science |
|---|---|---|---|
| BO1 | Reproduction des résultats scientifiques | DSO1 | Reproduire le Q-Learning de l'article (5 configurations) et comparer aux scores publiés |
| BO2 | Amélioration des performances de l'agent | DSO2 | Optimiser discrétisation, hyperparamètres et exploration |
| BO3 | Exploration de nouvelles approches | DSO3 | DQN, PPO (MountainCar-v0) et TD3 (MountainCarContinuous-v0) |
| BO4 | Évaluation comparative et sélection | DSO4 | Comparer performance, convergence, stabilité et coût avec un protocole reproductible |

### 1.4 Critères de succès

| BO | Critère |
|---|---|
| BO1 | Résultats du même ordre de grandeur que l'article (meilleur score publié : −108), écarts expliqués |
| BO2 | Récompense moyenne ≥ −110 et taux de réussite ≥ 90 % en évaluation (ε = 0), temps d'entraînement réduit |
| BO3 | DQN et PPO évalués sur MountainCar-v0 ; TD3 sur MountainCarContinuous-v0 |
| BO4 | Tableau comparatif (moyenne ± écart-type) et conclusion argumentée |

**Métriques** : récompense moyenne, taux de réussite, nombre moyen d'étapes, temps d'entraînement, convergence, stabilité (écart-type entre graines).

### 1.5 Protocole de comparaison

- 5 à 10 graines aléatoires par configuration, moyenne ± écart-type.
- Évaluation sans exploration (ε = 0) sur 100 épisodes distincts de l'entraînement.
- Même environnement et budget documenté pour toutes les méthodes.
- Pas de comparaison directe entre MountainCar-v0 et MountainCarContinuous-v0.

### 1.6 Risques

| Risque | Mesure |
|---|---|
| L'agent n'atteint jamais le drapeau au début | ε = 1 au départ, initialisation optimiste de Q |
| Ambiguïté du protocole de l'article | Documenter chaque choix d'implémentation |
| Forte variabilité entre graines | 5 à 10 graines, moyenne ± écart-type |
| Confusion terminated / truncated | Ne pas traiter la troncature comme un état terminal |
